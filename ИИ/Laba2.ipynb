```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv("Airline_Delay_Cause.csv")

df.head()
```

```python
df = df.fillna(0)
Q1 = df['arr_delay'].quantile(0.25)
Q3 = df['arr_delay'].quantile(0.75)
IQR = Q3 - Q1
df = df[(df['arr_delay'] >= Q1 - 1.5*IQR) & (df['arr_delay'] <= Q3 + 1.5*IQR)]
```

Исправлены замечания:
обработаны пропуски  
удалены выбросы 

```python
categorical_cols = ['carrier', 'airport']
df[categorical_cols].head()
```

Категориальные признаки включают:
авиакомпанию
аэропорт

```python
le = LabelEncoder()
df_le = df.copy()
df_le['carrier'] = le.fit_transform(df_le['carrier'])
df_le['airport'] = le.fit_transform(df_le['airport'])
```

```python
df_ohe = pd.get_dummies(df, columns=['carrier', 'airport'], drop_first=True)
df_ohe.head()
```

Использованы два метода кодирования:
Label Encoding
One-Hot Encoding

```python
num_cols = ['arr_flights', 'arr_delay', 'carrier_delay', 'weather_delay']
scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[num_cols] = scaler.fit_transform(df_scaled[num_cols])
df_scaled.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler()
df_minmax = df.copy()
df_minmax[num_cols] = minmax.fit_transform(df_minmax[num_cols])
df_minmax.head()
```

Сравнение методов масштабирования:
StandardScaler - нормализация к среднему
MinMaxScaler - приведение к диапазону

```python
df['delay_per_flight'] = df['arr_delay'] / (df['arr_flights'] + 1)
df['total_delay'] = (
    df['carrier_delay'] +
    df['weather_delay'] +
    df['nas_delay'] +
    df['late_aircraft_delay']
)
```

Созданы новые признаки:
задержка на один рейс
суммарные задержки по причинам

```python
df = df.drop(columns=[
    'carrier_name',
    'airport_name'
])
```

Удалены признаки:
carrier_name
airport_name
Причина - текстовые дубли категориальных признаков, не несут дополнительной информации
